```python
import random
import datetime
```

| `Event` | `HOME` | `AWAY` |
|---|---|---|
| Motion or door opens	|  nothing	| HIGH security |
| Smoke	| CRITICAL	| CRITICAL |
| Temp below 35°F or above 95°F	| safety alert	| safety alert |
| Temp outside 65–75°F | comfort notice | nothing |

```python
'''
def generate_reading(sensor):
    if sensor_type == "temperature":
        return random.randint(30, 100)
    elif sensor_type == "motion":
        return random.choice([True, False])

'''



class HomeSecurity:
    def __init__(self, time_log=None, mode="Home", door="Closed",alert_source=None,
                 temp_sensor=random.randint(20,120), security_status="Normal", smoke_sensor=False):
        self.time_log = datetime.datetime.now()
        self.mode = mode
        self.door = door
        self.temp_sensor = temp_sensor
        self.smoke_sensor = smoke_sensor
        self.security_status = security_status
        self.alert_source = alert_source



    def update_mode(self):

        if self.mode == "Away":
            if self.smoke_sensor:
                self.security_status = "Fire Alert"
                self.alert_source = ("Bedroom Smoke", self.smoke_sensor)

            if self.door == "Open":
                self.security_status = "High Security Alert"
                self.alert_source = ("Front door", self.door)

            if 35 < self.temp_sensor < 95:
                self.security_status = "Safety Alert"
                self.alert_source = ("Kitchen Temperature", self.temp_sensor)
            

        elif self.mode == "Home":
            
            if self.smoke_sensor:
                self.security_status = "Fire Alert"
                self.alert_source = ("Bedroom Smoke", self.smoke_sensor)
                
            if 95  <= self.temp_sensor or self.temp_sensor <= 35:
                self.security_status = "Safety Alert"
                self.alert_source = ("Kitchen Temperature", self.temp_sensor)

            if 65 <= self.temp_sensor <= 75:
                self.security_status = "Comfort"
         
         
        self.time_log = datetime.datetime.now()

    def display_status(self):
        if self.security_status == "Normal"or self.security_status == "Comfort":
            print("=== HomeGuard Security System ===")
            print("Time:", self.time_log)
            print("Mode:", self.mode)
            print("Security Status:", self.security_status)
            print("Front door:", self.door)
            print("Kitchen Temperature:", self.temp_sensor)
            print("Bedroom Smoke:", self.smoke_sensor)
            
        else:
            
            print("=== HomeGuard Security System ===")
            print("Time:", self.time_log)
            print("Mode:", self.mode)
            print("Security Status:", self.security_status)
            print("Alert Source:", self.alert_source[0], ":", self.alert_source[1])

        
```

```python
system = HomeSecurity(mode="Home", door="Open")
system.update_mode()
system.display_status()
```
